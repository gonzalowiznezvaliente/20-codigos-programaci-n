a) ¿Qué letras se imprimen y en qué orden?

**Ejercicio 2. Orden de los except**

```python
try:
  f = open("notas.txt")
  n = int(f.readline())
except Exception:
  print("Hubo un error")
except FileNotFoundError:
  print("El fichero no existe")
except ValueError:
  print("El contenido no es un número")
```

a) ¿Qué está mal?
El orden de las excepciones en los bloques except no es el correcto. Exception es la clase base general de la que heredan excepciones más específicas como FileNotFoundError y ValueError. En Python, las excepciones se evalúan de arriba hacia abajo. Como la primera línea es except Exception:, cualquier error (incluidos FileNotFoundError y ValueError) será capturado por esta primera cláusula, haciendo que las dos siguientes (FileNotFoundError y ValueError) sean código inalcanzable (unreachable code).   

Además, hay pequeños errores sintácticos en el código presentado:

- La variable n tiene asignada una conversión incompleta o con paréntesis (n = (f.readline())), aunque faltaría un float() o int() para que pudiese saltar el ValueError al leer la línea.   
- Hay un punto erróneo en except. FileNotFoundError:

b) ¿Cómo lo corregirías? Escribe el bloque corregido.

Se deben ordenar los bloques except desde las excepciones más específicas hasta la más general (Exception)

```python
try:
    f = open("notas.txt")
    n = int(f.readline())
except FileNotFoundError:
    print("El fichero no existe")
except ValueError:
    print("El contenido no es un número")
except Exception:
    print("Hubo un error")
```

**Ejercicio 3. else y finally con return**

```python
def misterio(n):
  try:
    print("1-try")
    if n == 0:
      return 0
    x = 10 / n
  except ZeroDivisionError:
    print("2-except")
    return -1
  else:
    print("3-else")
    return x
  finally:
    print("4-finally")

print("Resultado:", misterio(2))
print("---")
print("Resultado:", misterio(0))
```

b) Con misterio(2) y con misterio(0), ¿se ejecuta el else? ¿Y el finally?

Con misterio(2): Se ejecutan tanto el bloque else (porque no hay excepción) como el bloque finally.   

Con misterio(0): No se ejecuta el bloque else (porque la función sale antes con un return 0), pero sí se ejecuta el bloque finally (que se ejecuta siempre).

**Ejercicio 5. Corrige y completa**

```python
def pedir_nota():
  try:
    nota = int(input("Introduce nota (0-10, -1 para salir): "))
    if nota < -1 or nota > 10:
      raise ValueError("Nota fuera de rango")
  except ValueError:
    print("Valor no válido")
  else:
    return nota
  finally:
    return -2

while True:
  n = pedir_nota()
  if n == -1:
    break
print(f"Nota guardada: {n}")
```

a) Hay tres errores graves. Explica cada uno.

1. Condición incorrecta: if nota-1 or nota > 10: evalúa una resta aritmética en lugar de una igualdad. Debería ser if nota == -1: para salir.   
2. return en el bloque finally: Incluir return -2 en el finally provoca que la función siempre devuelva -2, sobrescribiendo cualquier otro valor retornado previamente en el try o el else.   
3. Falta de bucle para reintentos: Si se introduce un valor no válido, se captura la excepción y se ejecuta el finally, pero no se vuelve a pedir la nota.

b) Reescribe pedir_nota() correctamente.

```python
def pedir_nota():
    while True:
        try:
            nota = int(input("Introduce nota (0-10, -1 para salir): "))
            if nota == -1:
                return -1
            if nota < 0 or nota > 10:
                raise ValueError("Nota fuera de rango")
            return nota
        except ValueError:
            print("Valor no válido. Inténtalo de nuevo.")
```
